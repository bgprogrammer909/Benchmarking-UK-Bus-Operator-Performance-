# SECTION 6 – Process Disruption Dataset (SIRI-SX)

This section ingests, cleans, parses XML disruptions (SIRI-SX format), aggregates impact summaries by severity, and exports data artifacts.

### Cell 1 – Import Libraries

```python
import xml.etree.ElementTree as ET
import pandas as pd
from pyspark.sql.functions import col, count, lit, when
from pyspark import StorageLevel
```

### Cell 2 – Load XML File

```python
DISRUPTION_FILE = r"D:\big data assgn\sirisx.xml"
tree = ET.parse(DISRUPTION_FILE)
root = tree.getroot()
print("XML Root Tag:", root.tag)
```

### Cell 3 – Display XML Namespaces

```python
for child in root.iter():
    print("First Tag Found:", child.tag)
    break
```

### Cell 4 – Extract Disruption Records

```python
records = []
for element in root.iter():
    tag = element.tag.split("}")[-1]
    if tag == "PtSituationElement":
        record = {}
        for child in element:
            child_tag = child.tag.split("}")[-1]
            record[child_tag] = child.text
        records.append(record)
print("Disruptions Found:", len(records))
```

### Cell 5 – Create Pandas DataFrame

```python
disruption_df = pd.DataFrame(records)
disruption_df.head()
```

### Cell 6 – View Columns

```python
print("Extracted Columns:", disruption_df.columns.tolist())
```

### Cell 7 – Convert to Spark DataFrame

```python
disruption_spark = spark.createDataFrame(disruption_df)
disruption_spark.show(5, truncate=False)
```

### Cell 8 – Check Schema

```python
disruption_spark.printSchema()
```

### Cell 9 – Record Count

```python
print("Total Disruptions:", disruption_spark.count())
```

### Cell 10 – Remove Duplicate Records

```python
disruption_spark = disruption_spark.dropDuplicates()
print("Rows after deduplication:", disruption_spark.count())
```

### Cell 11 – Cache Dataset

```python
disruption_spark.cache()
disruption_spark.count()
```

### Cell 12 – Register SQL Table

```python
disruption_spark.createOrReplaceTempView("disruptions")
```

### Cell 13 – Show Available Columns

```python
print("Spark DF Columns:", disruption_spark.columns)
```

### Cell 14 – Create Disruption Summary

```python
severity_column = "Severity"
if severity_column not in disruption_spark.columns:
    disruption_spark = disruption_spark.withColumn("Severity", lit("Unknown"))

disruption_summary = (
    disruption_spark
    .groupBy("Severity")
    .agg(
        count("*").alias("TotalDisruptions")
    )
)
```

### Cell 15 – View Summary

```python
disruption_summary.show(truncate=False)
```

### Cell 16 – SQL Analysis

```python
disruption_summary.createOrReplaceTempView("disruption_summary")
spark.sql("""
SELECT *
FROM disruption_summary
ORDER BY TotalDisruptions DESC
""").show(truncate=False)
```

### Cell 17 – Null Value Check

```python
disruption_summary.select(
    [
        count(when(col(c).isNull(), c)).alias(c)
        for c in disruption_summary.columns
    ]
).show()
```

### Cell 18 – Summary Statistics

```python
disruption_summary.describe().show()
```

### Cell 19 – Check Partitions

```python
print("Initial Partitions:", disruption_summary.rdd.getNumPartitions())
```

### Cell 20 – Repartition

```python
disruption_summary = disruption_summary.repartition(8)
print("New Partitions:", disruption_summary.rdd.getNumPartitions())
```

### Cell 21 – Persist

```python
disruption_summary.persist(StorageLevel.MEMORY_AND_DISK)
disruption_summary.count()
```

### Cell 22 – Execution Plan

```python
disruption_summary.explain(extended=True)
```

### Cell 23 – Save as Parquet

```python
disruption_summary.write.mode("overwrite").parquet("output/disruption_summary")
```

### Cell 24 – Save as CSV

```python
disruption_summary.toPandas().to_csv(
    "output/disruption_summary.csv",
    index=False
)
```
