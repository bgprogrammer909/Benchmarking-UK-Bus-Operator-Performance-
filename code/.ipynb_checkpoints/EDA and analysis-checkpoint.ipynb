# Notebook 2 – Exploratory Data Analysis (EDA)

## SECTION 1 – Import Libraries

```python
import os
import sys
import pandas as pd
import matplotlib.pyplot as plt
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
```

## SECTION 2 – Create Spark Session

```python
os.environ["PYSPARK_PYTHON"] = sys.executable
os.environ["PYSPARK_DRIVER_PYTHON"] = sys.executable

spark = (
    SparkSession.builder
    .appName("Bus Operator Benchmarking - EDA")
    .config("spark.sql.shuffle.partitions", "8")
    .config("spark.driver.memory", "4g")
    .config("spark.executor.memory", "4g")
    .getOrCreate()
)

spark.sparkContext.setLogLevel("WARN")
print(spark.version)
```

## SECTION 3 – Load Final Dataset

```python
benchmark_df = spark.read.parquet("output/benchmark_dataset")
benchmark_df.cache()
benchmark_df.count()
```

## SECTION 4 – Dataset Overview

```python
print("Rows :", benchmark_df.count())
print("Columns :", len(benchmark_df.columns))
benchmark_df.printSchema()
benchmark_df.show(5, truncate=False)
```

## SECTION 5 – Summary Statistics

```python
benchmark_df.describe().show()
```

## SECTION 6 – Null Value Analysis

```python
benchmark_df.select(
    [
        count(
            when(
                col(c).isNull(),
                c
            )
        ).alias(c)
        for c in benchmark_df.columns
    ]
).show()
```

## SECTION 7 – Cardinality

```python
for c in benchmark_df.columns:
    print(
        c,
        benchmark_df.select(c).distinct().count()
    )
```

## SECTION 8 – Mean, Median, Standard Deviation

```python
benchmark_df.select(
    mean("TotalTrips").alias("MeanTrips"),
    expr("percentile_approx(TotalTrips,0.5)").alias("MedianTrips"),
    stddev("TotalTrips").alias("StdTrips"),
    mean("TotalStops").alias("MeanStops"),
    expr("percentile_approx(TotalStops,0.5)").alias("MedianStops"),
    stddev("TotalStops").alias("StdStops")
).show()
```

## SECTION 9 – Skewness & Kurtosis

```python
benchmark_df.select(
    skewness("TotalTrips"),
    kurtosis("TotalTrips"),
    skewness("TotalStops"),
    kurtosis("TotalStops")
).show()
```

## SECTION 10 – Correlation

```python
print(benchmark_df.stat.corr("TotalTrips", "TotalStops"))
```

## SECTION 11 – Largest Operators

```python
benchmark_df.orderBy(desc("PerformanceScore")).show(20, truncate=False)
```

## SECTION 12 – Smallest Operators

```python
benchmark_df.orderBy(asc("PerformanceScore")).show(20, truncate=False)
```

## SECTION 13 – SQL Analysis

```python
benchmark_df.createOrReplaceTempView("benchmark")
spark.sql("""
SELECT
Operator,
PerformanceScore,
TotalTrips,
TotalStops,
UniqueRoutes,
UniqueServices
FROM benchmark
ORDER BY PerformanceScore DESC
LIMIT 20
""").show(truncate=False)
```

## SECTION 14 – Convert to Pandas

```python
eda_df = benchmark_df.toPandas()
```

## SECTION 15 – Histogram

```python
plt.figure(figsize=(8, 5))
plt.hist(eda_df["TotalTrips"], bins=30)
plt.title("Distribution of Total Trips")
plt.xlabel("Trips")
plt.ylabel("Frequency")
plt.show()
```

## SECTION 16 – Scatter Plot

```python
plt.figure(figsize=(8, 6))
plt.scatter(eda_df["TotalTrips"], eda_df["TotalStops"])
plt.xlabel("Total Trips")
plt.ylabel("Total Stops")
plt.title("Trips vs Stops")
plt.show()
```

## SECTION 17 – Box Plot

```python
plt.figure(figsize=(6, 5))
plt.boxplot(eda_df["TotalTrips"])
plt.title("Total Trips")
plt.show()
```

## SECTION 18 – Top 20 Operators

```python
top20 = eda_df.sort_values("PerformanceScore", ascending=False).head(20)
plt.figure(figsize=(14, 6))
plt.bar(top20["Operator"], top20["PerformanceScore"])
plt.xticks(rotation=90)
plt.title("Top 20 Operators")
plt.show()
```

## SECTION 19 – Outlier Detection

```python
Q1 = eda_df["TotalTrips"].quantile(0.25)
Q3 = eda_df["TotalTrips"].quantile(0.75)
IQR = Q3 - Q1
outliers = eda_df[
    (eda_df["TotalTrips"] < (Q1 - 1.5 * IQR)) |
    (eda_df["TotalTrips"] > (Q3 + 1.5 * IQR))
]
print("Outliers :", len(outliers))
outliers.head()
```

## SECTION 20 – Cache & Execution Plan

```python
benchmark_df.cache()
benchmark_df.count()
benchmark_df.explain(True)
```

## SECTION 21 – Repartition

```python
benchmark_df = benchmark_df.repartition(8)
print(benchmark_df.rdd.getNumPartitions())
```

## SECTION 22 – Save EDA Dataset

```python
benchmark_df.write.mode("overwrite").parquet("output/eda_dataset")
```

## SECTION 23 – Export CSV

```python
benchmark_df.toPandas().to_csv("output/eda_dataset.csv", index=False)
```
