# SECTION 2 — Parse TransXChange XML Files

This section extracts service metadata (Service Code, Line Name, Route Description, Trip/Stop counts) from TransXChange XML files, processes the dataset using Pandas, cleans and loads it into PySpark, and analyzes top bus operators.

### Cell 1 — Define XML Namespace

```python
import xml.etree.ElementTree as ET

NAMESPACE = {
    "txc": "http://www.transxchange.org.uk/"
}
```

### Cell 2 — Create an Empty List

```python
records = []
```

### Cell 3 — Parse Every XML File

```python
from tqdm import tqdm
from pathlib import Path

for xml_file in tqdm(xml_files):
    try:
        tree = ET.parse(xml_file)
        root = tree.getroot()

        operator_folder = Path(xml_file).parent.name
        operator_name = operator_folder.rsplit("_", 1)[0]

        service_code = None
        line_name = None
        route_description = None

        service = root.find(".//txc:Service", NAMESPACE)
        if service is not None:
            service_code_elem = service.find("txc:ServiceCode", NAMESPACE)
            if service_code_elem is not None:
                service_code = service_code_elem.text

            line_elem = service.find(".//txc:LineName", NAMESPACE)
            if line_elem is not None:
                line_name = line_elem.text

            desc_elem = service.find(".//txc:Description", NAMESPACE)
            if desc_elem is not None:
                route_description = desc_elem.text

        vehicle_journeys = root.findall(".//txc:VehicleJourney", NAMESPACE)
        trip_count = len(vehicle_journeys)

        stop_points = root.findall(".//txc:AnnotatedStopPointRef", NAMESPACE)
        stop_count = len(stop_points)

        records.append({
            "Operator": operator_name,
            "ServiceCode": service_code,
            "LineName": line_name,
            "Description": route_description,
            "Trips": trip_count,
            "Stops": stop_count,
            "XML_File": str(xml_file)
        })
    except Exception:
        continue
```

### Cell 4 — Create a Pandas DataFrame

```python
records_df = pd.DataFrame(records)
records_df.head()
```

### Cell 5 — Convert to PySpark

```python
services_df = spark.createDataFrame(records_df)
services_df.show(5, truncate=False)
```

### Cell 6 — Check Schema

```python
services_df.printSchema()
```

### Cell 7 — Record Count

```python
print("Total Records:", services_df.count())
```

### Cell 8 — Remove Duplicates

```python
services_df = services_df.dropDuplicates()
```

### Cell 9 — Handle Missing Values

```python
services_df = services_df.fillna({
    "ServiceCode": "Unknown",
    "LineName": "Unknown",
    "Description": "Unknown"
})
```

### Cell 10 — Register Spark SQL View

```python
services_df.createOrReplaceTempView("services")
```

### Cell 11 — Top Operators by Number of Services

```python
spark.sql("""
SELECT
    Operator,
    COUNT(*) AS TotalServices
FROM services
GROUP BY Operator
ORDER BY TotalServices DESC
LIMIT 20
""").show()
```

### Cell 12 — Cache the DataFrame

```python
services_df.cache()
services_df.count()
```

# SECTION 3 – Feature Engineering Using PySpark

This section focuses on data cleaning, type casting, aggregating operator-level features (total trips, average stops, unique routes, etc.), running PySpark SQL queries, and exporting the results to Parquet and CSV formats.

### Cell 1 – Check Dataset

```python
services_df.printSchema()
services_df.show(5, truncate=False)
```

### Cell 2 – Count Total Records

```python
print("Total Services:", services_df.count())
```

### Cell 3 – Check Missing Values

```python
from pyspark.sql.functions import col, count, when

services_df.select([
    count(when(col(c).isNull(), c)).alias(c)
    for c in services_df.columns
]).show()
```

### Cell 4 – Remove Duplicate Records

```python
services_df = services_df.dropDuplicates()
print("Rows After Removing Duplicates:")
print(services_df.count())
```

### Cell 5 – Convert Trips and Stops to Integer

```python
from pyspark.sql.functions import col

services_df = services_df \
    .withColumn("Trips", col("Trips").cast("int")) \
    .withColumn("Stops", col("Stops").cast("int"))
```

### Cell 6 – Create Operator-Level Features

```python
from pyspark.sql.functions import *

operator_features = services_df.groupBy("Operator").agg(
    count("*").alias("TotalServices"),
    sum("Trips").alias("TotalTrips"),
    avg("Trips").alias("AverageTrips"),
    sum("Stops").alias("TotalStops"),
    avg("Stops").alias("AverageStops"),
    countDistinct("LineName").alias("UniqueRoutes"),
    countDistinct("ServiceCode").alias("UniqueServiceCodes")
)
```

### Cell 7 – View Features

```python
operator_features.show(20, truncate=False)
```

### Cell 8 – Sort by Largest Operators

```python
operator_features.orderBy(
    col("TotalTrips").desc()
).show(20, truncate=False)
```

### Cell 9 – Register Spark SQL Table

```python
operator_features.createOrReplaceTempView("operator_features")
```

### Cell 10 – Spark SQL Query

```python
spark.sql("""
SELECT
    Operator,
    TotalServices,
    TotalTrips,
    AverageTrips,
    TotalStops,
    AverageStops,
    UniqueRoutes
FROM operator_features
ORDER BY TotalTrips DESC
LIMIT 20
""").show(truncate=False)
```

### Cell 11 – Cache the Data

```python
operator_features.cache()
operator_features.count()
```

### Cell 12 – Check Number of Partitions

```python
print(operator_features.rdd.getNumPartitions())
```

### Cell 13 – Repartition

```python
operator_features = operator_features.repartition(8)
print(operator_features.rdd.getNumPartitions())
```

### Cell 14 – Save as Parquet

```python
operator_features.write.mode("overwrite").parquet("output/operator_features")
```

### Cell 15 – Save to CSV

```python
operator_features.toPandas().to_csv(
    "output/operator_features.csv",
    index=False
)
```

# SECTION 4 – Create Operator Summary Table

This section aggregates comprehensive operator statistics (min/max/avg trips and stops), calculates a baseline performance score, runs SQL evaluations, and persists the dataset to Parquet and CSV files.

### Cell 1 – Load Functions

```python
from pyspark.sql.functions import *
```

### Cell 2 – Verify Dataset

```python
print("Rows :", services_df.count())
print("Columns :", len(services_df.columns))
services_df.show(5, truncate=False)
```

### Cell 3 – Calculate Operator Statistics

```python
operator_summary = (
    services_df
    .groupBy("Operator")
    .agg(
        count("*").alias("TotalServices"),
        countDistinct("ServiceCode").alias("UniqueServices"),
        countDistinct("LineName").alias("UniqueRoutes"),
        sum("Trips").alias("TotalTrips"),
        avg("Trips").alias("AverageTrips"),
        max("Trips").alias("MaximumTrips"),
        min("Trips").alias("MinimumTrips"),
        sum("Stops").alias("TotalStops"),
        avg("Stops").alias("AverageStops"),
        max("Stops").alias("MaximumStops"),
        min("Stops").alias("MinimumStops")
    )
)
```

### Cell 4 – View Results

```python
operator_summary.show(20, truncate=False)
```

### Cell 5 – Sort by Largest Operators

```python
operator_summary.orderBy(
    desc("TotalTrips")
).show(20, truncate=False)
```

### Cell 6 – Add Performance Score
This isn't the final ML score—it gives us a useful feature for later analysis.

```python
operator_summary = (
    operator_summary
    .withColumn(
        "PerformanceScore",
        round(
            col("TotalTrips") * 0.40 +
            col("TotalStops") * 0.30 +
            col("UniqueRoutes") * 15 +
            col("UniqueServices") * 10,
            2
        )
    )
)
```

### Cell 7 – View Scores

```python
operator_summary.select(
    "Operator",
    "PerformanceScore"
).orderBy(
    desc("PerformanceScore")
).show(20, truncate=False)
```

### Cell 8 – Register Spark SQL Table

```python
operator_summary.createOrReplaceTempView("operator_summary")
```

### Cell 9 – SQL Analysis

```python
spark.sql("""
SELECT
    Operator,
    TotalServices,
    UniqueRoutes,
    TotalTrips,
    AverageTrips,
    TotalStops,
    AverageStops,
    PerformanceScore
FROM operator_summary
ORDER BY PerformanceScore DESC
LIMIT 20
""").show(truncate=False)
```

### Cell 10 – Cache DataFrame

```python
operator_summary.cache()
operator_summary.count()
```

### Cell 11 – Check Partitions

```python
print("Partitions :", operator_summary.rdd.getNumPartitions())
```

### Cell 12 – Repartition

```python
operator_summary = operator_summary.repartition(8)
print("Partitions :", operator_summary.rdd.getNumPartitions())
```

### Cell 13 – Save as Parquet

```python
operator_summary.write.mode("overwrite").parquet("output/operator_summary")
```

### Cell 14 – Save as CSV

```python
operator_summary.toPandas().to_csv(
    "output/operator_summary.csv",
    index=False
)
```

### Cell 15 – Display Summary Statistics

```python
operator_summary.describe().show()
```

# SECTION 5 – Process Vehicle Location Dataset

This section ingests GTFS-Realtime (protobuf) vehicle positioning data, extracts vehicle/trip attributes, converts the payload into PySpark DataFrames, calculates route-level speed and location statistics, and persists the outputs.

### Cell 1 – Install GTFS Library
Run this once to ensure dependencies for reading GTFS-Realtime Protobuf files are available.

```python
!pip install gtfs-realtime-bindings protobuf
```

### Cell 2 – Import Libraries

```python
from google.transit import gtfs_realtime_pb2
import pandas as pd
from pyspark.sql.functions import *
```

### Cell 3 – Load GTFS-RT File
Update `GTFS_PATH` if your file is in a different location.

```python
GTFS_PATH = r"D:\big data assgn\gtfsrt_2026-07-28_094937\gtfsrt.bin"
feed = gtfs_realtime_pb2.FeedMessage()
with open(GTFS_PATH, "rb") as f:
    feed.ParseFromString(f.read())
print("Total GTFS Entities:", len(feed.entity))
```

### Cell 4 – Extract Vehicle Records

```python
vehicle_records = []
for entity in feed.entity:
    if entity.HasField("vehicle"):
        vehicle = entity.vehicle
        vehicle_records.append({
            "VehicleID": vehicle.vehicle.id,
            "TripID": vehicle.trip.trip_id,
            "RouteID": vehicle.trip.route_id,
            "Latitude": vehicle.position.latitude,
            "Longitude": vehicle.position.longitude,
            "Bearing": vehicle.position.bearing,
            "Speed": vehicle.position.speed,
            "Timestamp": vehicle.timestamp
        })
print("Vehicle Records:", len(vehicle_records))
```

### Cell 5 – Create Pandas DataFrame

```python
vehicle_df = pd.DataFrame(vehicle_records)
vehicle_df.head()
```

### Cell 6 – Convert to PySpark

```python
vehicle_spark = spark.createDataFrame(vehicle_df)
vehicle_spark.show(5, truncate=False)
```

### Cell 7 – Check Schema

```python
vehicle_spark.printSchema()
```

### Cell 8 – Remove Duplicate Records

```python
vehicle_spark = vehicle_spark.dropDuplicates()
print("Vehicle Records:", vehicle_spark.count())
```

### Cell 9 – Cache Dataset

```python
vehicle_spark.cache()
vehicle_spark.count()
```

### Cell 10 – Register SQL Table

```python
vehicle_spark.createOrReplaceTempView("vehicle_locations")
```

### Cell 11 – Route Statistics

```python
spark.sql("""
SELECT
    RouteID,
    COUNT(*) AS VehicleCount,
    AVG(Speed) AS AverageSpeed,
    MAX(Speed) AS MaximumSpeed
FROM vehicle_locations
GROUP BY RouteID
ORDER BY VehicleCount DESC
LIMIT 20
""").show(truncate=False)
```

### Cell 12 – Create Vehicle Summary

```python
vehicle_summary = (
    vehicle_spark
    .groupBy("RouteID")
    .agg(
        count("*").alias("VehicleCount"),
        avg("Speed").alias("AverageSpeed"),
        max("Speed").alias("MaximumSpeed"),
        min("Speed").alias("MinimumSpeed"),
        avg("Latitude").alias("AverageLatitude"),
        avg("Longitude").alias("AverageLongitude"),
        avg("Bearing").alias("AverageBearing")
    )
)
```

### Cell 13 – View Summary

```python
vehicle_summary.show(20, truncate=False)
```

### Cell 14 – Register SQL Table

```python
vehicle_summary.createOrReplaceTempView("vehicle_summary")
```

### Cell 15 – SQL Analysis

```python
spark.sql("""
SELECT *
FROM vehicle_summary
ORDER BY VehicleCount DESC
LIMIT 20
""").show(truncate=False)
```

### Cell 16 – Check Partitions

```python
print("Partitions :", vehicle_summary.rdd.getNumPartitions())
```

### Cell 17 – Repartition

```python
vehicle_summary = vehicle_summary.repartition(8)
print("Partitions :", vehicle_summary.rdd.getNumPartitions())
```

### Cell 18 – Persist Dataset

```python
from pyspark import StorageLevel
vehicle_summary.persist(StorageLevel.MEMORY_AND_DISK)
vehicle_summary.count()
```

### Cell 19 – Execution Plan

```python
vehicle_summary.explain(True)
```

```python
vehicle_summary.select(
    [
        count(when(col(c).isNull(), c)).alias(c)
        for c in vehicle_summary.columns
    ]
).show()
```

### Cell 21 – Summary Statistics

```python
vehicle_summary.describe().show()
```

### Cell 22 – Save as Parquet

```python
vehicle_summary.write.mode("overwrite").parquet("output/vehicle_summary")
```

### Cell 23 – Save as CSV

```python
vehicle_summary.toPandas().to_csv(
    "output/vehicle_summary.csv",
    index=False
)
```
