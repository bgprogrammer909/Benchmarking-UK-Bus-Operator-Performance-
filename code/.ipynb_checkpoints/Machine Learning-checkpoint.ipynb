# NOTEBOOK 3 — Machine Learning: Bus Operator Benchmarking

## SECTION 1 – Load Libraries & Spark

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
```

```python
spark = (
    SparkSession.builder
    .appName("Bus Operator Benchmarking - ML")
    .config("spark.sql.shuffle.partitions", "8")
    .getOrCreate()
)
spark.sparkContext.setLogLevel("WARN")
```

## SECTION 2 – Load Final Dataset

```python
benchmark_df = spark.read.parquet("output/benchmark_dataset")
benchmark_df.show(5)
```

```python
benchmark_df.printSchema()
```

```python
print("Operators :", benchmark_df.count())
```

## SECTION 3 – Feature Selection

```python
benchmark_df.select(
    "Operator", "TotalTrips", "TotalStops", "UniqueRoutes", 
    "PerformanceScore", "NetworkVehicleCount", "DisruptionCount"
).show(5)
```

## SECTION 4 – Assemble Features

```python
from pyspark.ml.feature import VectorAssembler
```

```python
assembler = VectorAssembler(
    inputCols=[
        "TotalTrips", "TotalStops", "UniqueRoutes", 
        "PerformanceScore", "NetworkVehicleCount", "DisruptionCount"
    ],
    outputCol="features"
)
ml_df = assembler.transform(benchmark_df)
```

```python
ml_df.select("Operator", "features").show(5, truncate=False)
```

## SECTION 5 – Feature Scaling

```python
from pyspark.ml.feature import StandardScaler
```

```python
scaler = StandardScaler(
    inputCol="features", 
    outputCol="scaledFeatures", 
    withMean=True, 
    withStd=True
)
scalerModel = scaler.fit(ml_df)
scaled_df = scalerModel.transform(ml_df)
```

```python
scaled_df.select("Operator", "scaledFeatures").show(5, truncate=False)
```

## SECTION 6 – K-Means Clustering

```python
from pyspark.ml.clustering import KMeans
```

```python
kmeans = KMeans(k=4, seed=42, featuresCol="scaledFeatures")
kmeans_model = kmeans.fit(scaled_df)
kmeans_df = kmeans_model.transform(scaled_df)
```

```python
kmeans_df.select("Operator", "prediction").show(20)
```

```python
kmeans_df.groupBy("prediction").count().show()
```

```python
from pyspark.ml.evaluation import ClusteringEvaluator
evaluator = ClusteringEvaluator(featuresCol="scaledFeatures")
score = evaluator.evaluate(kmeans_df)
print("Silhouette Score :", score)
```

## SECTION 7 – Bisecting K-Means

```python
from pyspark.ml.clustering import BisectingKMeans
```

```python
bkm = BisectingKMeans(k=4, featuresCol="scaledFeatures")
bkm_model = bkm.fit(scaled_df)
bkm_df = bkm_model.transform(scaled_df)
```

```python
bkm_df.groupBy("prediction").count().show()
```

```python
score = evaluator.evaluate(bkm_df)
print("Bisecting KMeans Score :", score)
```

## SECTION 8 – Gaussian Mixture

```python
from pyspark.ml.clustering import GaussianMixture
```

```python
gm = GaussianMixture(k=4, featuresCol="scaledFeatures")
gm_model = gm.fit(scaled_df)
gm_df = gm_model.transform(scaled_df)
```

```python
gm_df.select("Operator", "prediction").show(20)
```

```python
gm_df.groupBy("prediction").count().show()
```

## SECTION 9 – Compare Algorithms

```python
kmeans_score = evaluator.evaluate(kmeans_df)
bisect_score = evaluator.evaluate(bkm_df)
gm_score = evaluator.evaluate(gm_df)

comparison = spark.createDataFrame([
    ("KMeans", kmeans_score),
    ("Bisecting KMeans", bisect_score),
    ("Gaussian Mixture", gm_score)
], ["Algorithm", "SilhouetteScore"])

comparison.show()
```

```python
comparison.toPandas()
```

## SECTION 10 – Save Results

```python
kmeans_df.write.mode("overwrite").parquet("output/kmeans_results")
bkm_df.write.mode("overwrite").parquet("output/bisecting_results")
gm_df.write.mode("overwrite").parquet("output/gaussian_results")
```

```python
comparison.write.mode("overwrite").parquet("output/model_comparison")
```

## SECTION 11 – Visualize Clusters

```python
import matplotlib.pyplot as plt

cluster_df = kmeans_df.toPandas()
plt.figure(figsize=(10, 7))

for c in sorted(cluster_df["prediction"].unique()):
    subset = cluster_df[cluster_df["prediction"] == c]
    plt.scatter(
        subset["TotalTrips"],
        subset["PerformanceScore"],
        label=f"Cluster {c}"
    )

plt.xlabel("Total Trips")
plt.ylabel("Performance Score")
plt.title("K-Means Clustering of Bus Operators")
plt.legend()
plt.show()
```
