# SECTION 2 — Parse TransXChange XML Files

This section extracts service metadata (Service Code, Line Name, Route Description, Trip/Stop counts) from TransXChange XML files, processes the dataset using Pandas, cleans and loads it into PySpark, and analyzes top bus operators.

### Cell 1 — Define XML Namespace

```python
import xml.etree.ElementTree as ET

NAMESPACE = {
    "txc": "http://www.transxchange.org.uk/"
}
```

### Cell 2 — Create an Empty List

```python
records = []
```

### Cell 3 — Parse Every XML File

```python
from tqdm import tqdm
from pathlib import Path

for xml_file in tqdm(xml_files):
    try:
        tree = ET.parse(xml_file)
        root = tree.getroot()

        operator_folder = Path(xml_file).parent.name
        operator_name = operator_folder.rsplit("_", 1)[0]

        service_code = None
        line_name = None
        route_description = None

        service = root.find(".//txc:Service", NAMESPACE)
        if service is not None:
            service_code_elem = service.find("txc:ServiceCode", NAMESPACE)
            if service_code_elem is not None:
                service_code = service_code_elem.text

            line_elem = service.find(".//txc:LineName", NAMESPACE)
            if line_elem is not None:
                line_name = line_elem.text

            desc_elem = service.find(".//txc:Description", NAMESPACE)
            if desc_elem is not None:
                route_description = desc_elem.text

        vehicle_journeys = root.findall(".//txc:VehicleJourney", NAMESPACE)
        trip_count = len(vehicle_journeys)

        stop_points = root.findall(".//txc:AnnotatedStopPointRef", NAMESPACE)
        stop_count = len(stop_points)

        records.append({
            "Operator": operator_name,
            "ServiceCode": service_code,
            "LineName": line_name,
            "Description": route_description,
            "Trips": trip_count,
            "Stops": stop_count,
            "XML_File": str(xml_file)
        })
    except Exception:
        continue
```

### Cell 4 — Create a Pandas DataFrame

```python
records_df = pd.DataFrame(records)
records_df.head()
```

### Cell 5 — Convert to PySpark

```python
services_df = spark.createDataFrame(records_df)
services_df.show(5, truncate=False)
```

### Cell 6 — Check Schema

```python
services_df.printSchema()
```

### Cell 7 — Record Count

```python
print("Total Records:", services_df.count())
```

### Cell 8 — Remove Duplicates

```python
services_df = services_df.dropDuplicates()
```

### Cell 9 — Handle Missing Values

```python
services_df = services_df.fillna({
    "ServiceCode": "Unknown",
    "LineName": "Unknown",
    "Description": "Unknown"
})
```

### Cell 10 — Register Spark SQL View

```python
services_df.createOrReplaceTempView("services")
```

### Cell 11 — Top Operators by Number of Services

```python
spark.sql("""
SELECT
    Operator,
    COUNT(*) AS TotalServices
FROM services
GROUP BY Operator
ORDER BY TotalServices DESC
LIMIT 20
""").show()
```
