## Dataset Location

## Automatically Extract Every ZIP

```python
zip_files = list(Path(BASE_PATH).rglob("*.zip"))

print(f"ZIP files found : {len(zip_files)}")

for file in tqdm(zip_files):

    try:

        with zipfile.ZipFile(file,'r') as zip_ref:

            zip_ref.extractall(file.parent)

    except Exception as e:

        print(file)
```

## Find Every XML File

```python
xml_files = list(Path(BASE_PATH).rglob("*.xml"))

print("Total XML Files :", len(xml_files))
```

## Show Sample Files

```python
for file in xml_files[:10]:

    print(file)
```

## Count Operators

```python
operators = {}

for file in xml_files:

    operator = file.parent.name

    operators[operator] = operators.get(operator,0)+1
```

## Convert to Pandas

```python
operator_df = pd.DataFrame(
    operators.items(),
    columns=[
        "Operator",
        "Timetable Files"
    ]
)

operator_df.head()
```

## Convert to PySpark

```python
operator_spark = spark.createDataFrame(operator_df)

operator_spark.show(10)
```

# Feature Engineering Using PySpark

## Check Dataset

```python
services_df.printSchema()

services_df.show(5, truncate=False)
```

## Count Total Records

```python
print("Total Services:", services_df.count())
```

## Remove Duplicate Records

```python
services_df = services_df.dropDuplicates()

print("Rows After Removing Duplicates:")
print(services_df.count())
```

## Convert Trips and Stops to Integer

```python
from pyspark.sql.functions import col

services_df = services_df.withColumn("Trips", col("Trips").cast("int")).withColumn("Stops", col("Stops").cast("int"))
```

## Create Operator-Level Features

```python
from pyspark.sql.functions import *
from pyspark.sql.functions import (
    sum,
    avg,
    countDistinct
)

operator_features = services_df.groupBy("Operator").agg(

    count("*").alias("TotalServices"),

    sum("Trips").alias("TotalTrips"),

    avg("Trips").alias("AverageTrips"),

    sum("Stops").alias("TotalStops"),

    avg("Stops").alias("AverageStops"),

    countDistinct("LineName").alias("UniqueRoutes"),

    countDistinct("ServiceCode").alias("UniqueServiceCodes")

)
```

## View Features

```python
operator_features.show(20, truncate=False)
```

## Sort by Largest Operators

```python
operator_features.orderBy(
    col("TotalTrips").desc()
).show(20, truncate=False)
```

## Register Spark SQL Table

```python
operator_features.createOrReplaceTempView("operator_features")
```

## Spark SQL Query

```python
spark.sql("""

SELECT

Operator,

TotalServices,

TotalTrips,

AverageTrips,

TotalStops,

AverageStops,

UniqueRoutes

FROM operator_features

ORDER BY TotalTrips DESC

LIMIT 20

""").show(truncate=False)
```

## Cache the Data

```python
operator_features.cache()

operator_features.count()
```

## Check Number of Partitions

```python
print(operator_features.rdd.getNumPartitions())
```

## Repartition

```python
operator_features = operator_features.repartition(8)

print(operator_features.rdd.getNumPartitions())
```

## Save as Parquet

```python
operator_features.write.mode("overwrite").parquet("output/operator_features")
```

## Save to CSV

```python
operator_features.toPandas().to_csv(
    "output/operator_features.csv",
    index=False
)
```

## Read the GTFS-RT Filem

```python
from google.transit import gtfs_realtime_pb2
import pandas as pd
feed = gtfs_realtime_pb2.FeedMessage()

with open(r"D:\big data assgn\gtfsrt_2026-07-28_094937\gtfsrt.bin", "rb") as f:
    feed.ParseFromString(f.read())
print("Entities:", len(feed.entity))
```

## Extract Vehicle Positions

```python
vehicle_records = []

for entity in feed.entity:

    if entity.HasField("vehicle"):

        vehicle = entity.vehicle

        vehicle_records.append({

            "VehicleID": vehicle.vehicle.id,

            "TripID": vehicle.trip.trip_id,

            "RouteID": vehicle.trip.route_id,

            "Latitude": vehicle.position.latitude,

            "Longitude": vehicle.position.longitude,

            "Bearing": vehicle.position.bearing,

            "Speed": vehicle.position.speed,

            "Timestamp": vehicle.timestamp

        })
```

## Create Pandas DataFrame

```python
vehicle_df = pd.DataFrame(vehicle_records)

vehicle_df.head()
```

## Convert to PySpark

```python
vehicle_spark = spark.createDataFrame(vehicle_df)

vehicle_spark.show(10)
```

## Schema

```python
vehicle_spark.printSchema()
```

## Record Count

```python
print("Vehicle Records:", vehicle_spark.count())
```

## Remove Duplicates

```python
vehicle_spark = vehicle_spark.dropDuplicates()
```

## Cache

```python
vehicle_spark.cache()

vehicle_spark.count()
```

## Register SQL Table

```python
vehicle_spark.createOrReplaceTempView("vehicle_locations")
```

```python
spark.sql("""

SELECT

RouteID,

COUNT(*) AS Vehicles

FROM vehicle_locations

GROUP BY RouteID

ORDER BY Vehicles DESC

LIMIT 20

""").show()
```

## Save as Parquet

```python
vehicle_spark.write.mode("overwrite").parquet("output/vehicle_locations")
```
