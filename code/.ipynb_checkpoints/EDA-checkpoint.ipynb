# SECTION 7 – Build Final Benchmark Dataset

### Cell 1 – Import Functions

```python
from pyspark.sql.functions import *
```

### Cell 2 – Calculate Vehicle KPIs

```python
vehicle_kpis = vehicle_spark.agg(
    count("*").alias("NetworkVehicleCount"),
    avg("Speed").alias("NetworkAverageSpeed"),
    max("Speed").alias("NetworkMaximumSpeed"),
    min("Speed").alias("NetworkMinimumSpeed"),
    countDistinct("RouteID").alias("ActiveRoutes")
)
vehicle_kpis.show()
```

### Cell 3 – Calculate Disruption KPIs

```python
disruption_kpis = disruption_spark.agg(
    count("*").alias("NetworkDisruptions"),
    countDistinct("ParticipantRef").alias("PublishingAuthorities"),
    countDistinct("MiscellaneousReason").alias("UniqueDisruptionReasons")
)
disruption_kpis.show()
```

### Cell 4 – Convert KPI Tables to Python Dictionaries

```python
vehicle_values = vehicle_kpis.first().asDict()
disruption_values = disruption_kpis.first().asDict()
print(vehicle_values)
print(disruption_values)
```

### Cell 5 – Add Vehicle Features

```python
benchmark_df = (
    operator_summary
    .withColumn(
        "NetworkVehicleCount",
        lit(vehicle_values["NetworkVehicleCount"])
    )
    .withColumn(
        "NetworkAverageSpeed",
        lit(vehicle_values["NetworkAverageSpeed"])
    )
    .withColumn(
        "NetworkMaximumSpeed",
        lit(vehicle_values["NetworkMaximumSpeed"])
    )
    .withColumn(
        "NetworkMinimumSpeed",
        lit(vehicle_values["NetworkMinimumSpeed"])
    )
    .withColumn(
        "ActiveRoutes",
        lit(vehicle_values["ActiveRoutes"])
    )
)
```

### Cell 6 – Add Disruption Features

```python
benchmark_df = (
    benchmark_df
    .withColumn(
        "NetworkDisruptions",
        lit(disruption_values["NetworkDisruptions"])
    )
    .withColumn(
        "PublishingAuthorities",
        lit(disruption_values["PublishingAuthorities"])
    )
    .withColumn(
        "UniqueDisruptionReasons",
        lit(disruption_values["UniqueDisruptionReasons"])
    )
)
```

### Cell 7 – View Final Dataset

```python
benchmark_df.show(20, truncate=False)
```

### Cell 8 – Check Schema

```python
benchmark_df.printSchema()
```

### Cell 9 – Record Count

```python
print("Operators :", benchmark_df.count())
```

### Cell 10 – Register SQL Table

```python
benchmark_df.createOrReplaceTempView("benchmark")
```

### Cell 11 – SQL Analysis

```python
spark.sql("""
SELECT
    Operator,
    PerformanceScore,
    TotalTrips,
    TotalStops,
    UniqueRoutes,
    NetworkVehicleCount,
    NetworkAverageSpeed,
    NetworkDisruptions
FROM benchmark
ORDER BY PerformanceScore DESC
LIMIT 20
""").show(truncate=False)
```

### Cell 12 – Check Missing Values

```python
benchmark_df.select(
    [
        count(when(col(c).isNull(), col(c))).alias(c)
        for c in benchmark_df.columns
    ]
).show()
```

### Cell 13 – Summary Statistics

```python
benchmark_df.describe().show()
```

### Cell 14 – Persist Dataset

```python
from pyspark import StorageLevel

benchmark_df.persist(StorageLevel.MEMORY_AND_DISK)
benchmark_df.count()
```

### Cell 15 – Execution Plan

```python
benchmark_df.explain(True)
```

### Cell 16 – Repartition

```python
benchmark_df = benchmark_df.repartition(8)
print("Partitions :", benchmark_df.rdd.getNumPartitions())
```

### Cell 17 – Save as Parquet

```python
benchmark_df.write.mode("overwrite").parquet("output/benchmark_dataset")
```

### Cell 18 – Save as CSV

```python
benchmark_df.toPandas().to_csv(
    "output/benchmark_dataset.csv",
    index=False
)
```
