# 📊 Bus Operator Performance Benchmarking using Big Data Analytics and Machine Learning
## Notebook 1 of 4 — Data Engineering

**Stakeholder:** Government Transport Regulator (Great Britain)

**Business Problem**

Government regulators need to benchmark bus operators across Great Britain using publicly available
[Bus Open Data Service (BODS)](https://www.bus-data.dft.gov.uk/) datasets. This system identifies
high-performing and poor-performing operators by combining:

- **Timetable data** (TransXChange XML) — service structure, trips, stops
- **Vehicle location data** (GTFS-Realtime protobuf) — live operational behaviour
- **Disruption data** (SIRI-SX XML) — service reliability incidents

Rather than supervised learning (no ground-truth "good/bad operator" labels exist), we use
**unsupervised clustering** in Notebook 3 to group operators by operational behaviour.

---

### ⚠️ Note on data source for this notebook

No live BODS archive was supplied to this environment, so this notebook includes a
**synthetic data generator (Section 1)** that produces files in the **exact schema and format**
BODS publishes — real TransXChange XML, real GTFS-Realtime protobuf `FeedMessage` binaries, and
real SIRI-SX XML. Every parsing, cleaning, Spark and SQL cell below operates on these generated
files exactly as it would on a real `bodds_archive_20260727/` export — **to point this notebook at
real BODS data, change `RAW_DATA_DIR` in Section 1 to your archive path and skip the generator
cells.** Nothing downstream needs to change.

## 3. Synthetic BODS Data Generation

BODS publishes data per-operator. We simulate a realistic slice of the GB bus market: a mix of
large national groups (which run many services) and smaller independent/regional operators (which
run few), because that skew is exactly the kind of heterogeneity the clustering stage in
Notebook 3 needs to find real structure in.

Each operator gets:
- A **TransXChange XML** file per bus service (route)
- A rolling set of **GTFS-Realtime** protobuf snapshots (vehicle positions over time)
- Zero or more **SIRI-SX** disruption records


```python
# ----------------------------------------------------------------------
# Operator roster: mix of large groups (many services, many vehicles)
# and small independents (few services), to create real behavioural variance.
# ----------------------------------------------------------------------
OPERATOR_GROUPS = {
    "Stagecoach North East": 9, "Stagecoach South": 7, "Stagecoach Midlands": 8,
    "First Bus West of England": 8, "First Bus Yorkshire": 6, "First Bus Essex": 5,
    "Arriva Midlands": 9, "Arriva North West": 8, "Arriva Southern Counties": 6,
    "Go-Ahead London": 10, "Go North East": 7, "Brighton & Hove Buses": 6,
    "National Express West Midlands": 9, "Reading Buses": 5, "Lothian Buses": 8,
    "Translink Ulsterbus": 6, "Nottingham City Transport": 6, "Trent Barton": 5,
    "Blackpool Transport": 4, "Cardiff Bus": 5, "Ipswich Buses": 3,
    "Warrington's Own Buses": 3, "Metrobus": 4, "Salisbury Reds": 2,
    "Hedingham Buses": 2, "Compass Travel": 2, "Dartline Coaches": 2,
    "Selwyns Travel": 1, "Yorkshire Coastliner": 2, "Konectbus": 2,
}

# Deterministic OperatorID per operator (mirrors how a regulator would key operators)
OPERATOR_IDS = {name: f"OP{idx+1:03d}" for idx, name in enumerate(sorted(OPERATOR_GROUPS))}

TOWN_NAMES = [fake.city() for _ in range(60)]
print(f"{len(OPERATOR_GROUPS)} operators configured. Total services to generate:",
      sum(OPERATOR_GROUPS.values()))
```

```python
def build_transxchange_xml(operator, operator_id, service_code, line_name, num_trips, num_stops):
    """
    Builds a real (schema-valid, simplified) TransXChange document string.
    TransXChange is the UK national standard XML schema for bus schedules.
    """
    ns = "http://www.transxchange.org.uk/"
    stop_points = TOWN_NAMES[:num_stops] if num_stops <= len(TOWN_NAMES) else (TOWN_NAMES * 3)[:num_stops]

    stop_xml = "".join(
        f'<AnnotatedStopPointRef><StopPointRef>ATC{operator_id}{i:03d}</StopPointRef>'
        f'<CommonName>{stop}</CommonName></AnnotatedStopPointRef>'
        for i, stop in enumerate(stop_points)
    )

    journeys_xml = "".join(
        f'<VehicleJourney><VehicleJourneyCode>VJ{operator_id}{service_code}{t:03d}</VehicleJourneyCode>'
        f'<ServiceRef>{service_code}</ServiceRef><LineRef>{line_name}</LineRef>'
        f'<DepartureTime>{6 + (t % 16):02d}:{(t * 7) % 60:02d}:00</DepartureTime></VehicleJourney>'
        for t in range(num_trips)
    )

    description = f"{stop_points[0]} to {stop_points[-1]} via {len(stop_points)} stops"

    xml_str = f"""<?xml version="1.0" encoding="UTF-8"?>
<TransXChange xmlns="{ns}" CreationDateTime="2026-07-27T00:00:00" SchemaVersion="2.4">
  <Operators>
    <Operator id="{operator_id}">
      <OperatorCode>{operator_id}</OperatorCode>
      <OperatorShortName>{operator}</OperatorShortName>
    </Operator>
  </Operators>
  <StopPoints>{stop_xml}</StopPoints>
  <Services>
    <Service>
      <ServiceCode>{service_code}</ServiceCode>
      <Lines><Line id="{line_name}"><LineName>{line_name}</LineName></Line></Lines>
      <Description>{description}</Description>
      <RegisteredOperatorRef>{operator_id}</RegisteredOperatorRef>
    </Service>
  </Services>
  <VehicleJourneys>{journeys_xml}</VehicleJourneys>
</TransXChange>"""
    return xml_str, description
```

```python
# ----------------------------------------------------------------------
# Write the TransXChange archive: one folder per operator (mirrors real BODS export)
# ----------------------------------------------------------------------
if os.path.exists(TIMETABLE_DIR):
    shutil.rmtree(TIMETABLE_DIR)
os.makedirs(TIMETABLE_DIR, exist_ok=True)

service_registry = []  # used later to generate consistent GTFS-RT / disruption data

for operator, n_services in OPERATOR_GROUPS.items():
    operator_id = OPERATOR_IDS[operator]
    op_folder = os.path.join(TIMETABLE_DIR, operator.replace(" ", "_").replace("&", "and"))
    os.makedirs(op_folder, exist_ok=True)

    # Larger operators run busier, longer routes; smaller ones run leaner services
    scale = min(1.0, n_services / 10)
    for s in range(n_services):
        line_name = f"{random.choice(['X', ''])}{random.randint(1, 99)}"
        service_code = f"{operator_id}-{s+1:02d}"
        num_trips = int(np.clip(np.random.normal(20 + 25 * scale, 6), 4, 60))
        num_stops = int(np.clip(np.random.normal(8 + 10 * scale, 3), 3, 30))

        xml_str, description = build_transxchange_xml(
            operator, operator_id, service_code, line_name, num_trips, num_stops
        )
        file_path = os.path.join(op_folder, f"{service_code}.xml")
        with open(file_path, "w", encoding="utf-8") as f:
            f.write(xml_str)

        service_registry.append({
            "Operator": operator, "OperatorID": operator_id, "ServiceCode": service_code,
            "LineName": line_name, "NumTrips": num_trips, "NumStops": num_stops,
        })

print(f"Generated {len(service_registry)} TransXChange XML files across "
      f"{len(OPERATOR_GROUPS)} operator folders.")
print("Example folder:", os.path.relpath(op_folder, PROJECT_ROOT))
```

## 4. Load & Parse Timetable XML (Distributed with PySpark)

BODS timetable exports contain **thousands of small XML files** — one per operator per service —
which is a classic "many small files" big-data pattern. Rather than pull in the `spark-xml` Maven
package (which needs internet access to a Maven repo not available here), we parse using a
**Spark-native distributed pattern**:

1. List every file path (driver-side, cheap — just metadata)
2. `sc.parallelize()` the *list of paths* across the cluster (not the file contents)
3. Each executor task independently opens and parses its assigned files with `xml.etree.ElementTree`
4. `flatMap` back a list of `Row` dicts and build a typed Spark DataFrame

This scales linearly with executors/cores and avoids driver-side I/O bottlenecks, which is exactly
how you'd approach this on a real multi-node cluster with the true BODS archive.


```python
xml_file_paths = glob.glob(os.path.join(TIMETABLE_DIR, "*", "*.xml"))
print(f"Discovered {len(xml_file_paths)} timetable XML files to parse.")

# Distribute the *list of file paths* across partitions — this is the key big-data pattern
# for "many small files": parallelism happens over files, not over bytes within one file.
NUM_PARSE_PARTITIONS = min(8, len(xml_file_paths))
paths_rdd = spark.sparkContext.parallelize(xml_file_paths, numSlices=NUM_PARSE_PARTITIONS)
print("Partitions for XML parsing stage:", paths_rdd.getNumPartitions())
```

```python
def parse_transxchange_file(file_path):
    """Runs on the executor for each assigned file. Returns a list of plain dict rows."""
    ns = {"tx": "http://www.transxchange.org.uk/"}
    try:
        tree = ET.parse(file_path)
        root = tree.getroot()

        operator_el = root.find(".//tx:Operators/tx:Operator/tx:OperatorShortName", ns)
        service_el = root.find(".//tx:Services/tx:Service", ns)
        service_code = service_el.find("tx:ServiceCode", ns).text
        line_name = service_el.find(".//tx:Line/tx:LineName", ns).text
        description = service_el.find("tx:Description", ns).text

        trips = len(root.findall(".//tx:VehicleJourneys/tx:VehicleJourney", ns))
        stops = len(root.findall(".//tx:StopPoints/tx:AnnotatedStopPointRef", ns))

        return [{
            "Operator": operator_el.text,
            "ServiceCode": service_code,
            "LineName": line_name,
            "Description": description,
            "Trips": trips,
            "Stops": stops,
            "XML_File": file_path,
        }]
    except Exception as e:
        # Malformed-file resilience: log and skip rather than kill the whole job
        return [{
            "Operator": None, "ServiceCode": None, "LineName": None,
            "Description": f"PARSE_ERROR: {e}", "Trips": None, "Stops": None,
            "XML_File": file_path,
        }]

parsed_rows_rdd = paths_rdd.flatMap(parse_transxchange_file)
print("Sample parsed row:", parsed_rows_rdd.first())
```

```python
timetable_schema = T.StructType([
    T.StructField("Operator", T.StringType(), True),
    T.StructField("ServiceCode", T.StringType(), True),
    T.StructField("LineName", T.StringType(), True),
    T.StructField("Description", T.StringType(), True),
    T.StructField("Trips", T.IntegerType(), True),
    T.StructField("Stops", T.IntegerType(), True),
    T.StructField("XML_File", T.StringType(), True),
])

timetable_df = spark.createDataFrame(parsed_rows_rdd, schema=timetable_schema)
print("Rows parsed:", timetable_df.count())
timetable_df.show(5, truncate=40)
```

## 5. Caching, Partitioning & Execution-Plan Analysis

Since `timetable_df` will be reused across several downstream aggregations (Operator Summary,
join with vehicle/disruption data, SQLite export), we **cache** it to avoid recomputing the XML
parse from scratch on every action. We also **repartition by `Operator`** so that later
operator-level `groupBy`/joins avoid an extra shuffle.


```python
print("Partitions before repartition:", timetable_df.rdd.getNumPartitions())

timetable_df = timetable_df.repartition(8, "Operator")
timetable_df = timetable_df.persist(StorageLevel.MEMORY_AND_DISK)
timetable_df.count()  # materialise the cache (Spark is lazy — actions trigger execution)

print("Partitions after repartition:", timetable_df.rdd.getNumPartitions())
print("Is cached:", timetable_df.is_cached)
```

```python
# ----------------------------------------------------------------------
# Lazy evaluation & explain-plan discussion
# ----------------------------------------------------------------------
# Nothing above actually ran the parse until .count() was called — transformations
# (repartition, filters, selects) build a logical plan; only actions (count, show,
# collect, write) trigger a physical execution ("lazy evaluation").
example_transform = timetable_df.filter(F.col("Trips") > 15).groupBy("Operator").agg(
    F.count("*").alias("num_services"), F.avg("Trips").alias("avg_trips")
)

print("Logical + physical plan for a representative downstream query:\n")
example_transform.explain(mode="formatted")
```

**Reading the plan above:** the `Filter` and `HashAggregate` stages operate within existing
partitions where possible; Spark's Adaptive Query Execution (AQE) — enabled in our Spark config —
can also coalesce small shuffle partitions at runtime. You can inspect live DAGs, stages and task
skew for any action in the **Spark UI** (`spark.sparkContext.uiWebUrl`, e.g. `http://localhost:4040`
during a running local session) — the *Jobs*, *Stages* and *SQL / DataFrame* tabs are the most
useful for diagnosing shuffle cost and partition skew on a real cluster run.

## 6. Data Cleaning

```python
rows_before = timetable_df.count()

timetable_clean_df = (
    timetable_df
    .filter(F.col("Operator").isNotNull())                     # drop unparseable files
    .withColumn("Operator", F.trim(F.col("Operator")))
    .withColumn("LineName", F.trim(F.col("LineName")))
    .withColumn("Description", F.trim(F.col("Description")))
    .filter((F.col("Trips") > 0) & (F.col("Stops") > 1))        # a "service" needs both
    .dropDuplicates(["ServiceCode"])
)

# Swap the cache: unpersist the raw version now that the clean version supersedes it
timetable_df.unpersist()
timetable_clean_df = timetable_clean_df.persist(StorageLevel.MEMORY_AND_DISK)
rows_after = timetable_clean_df.count()

print(f"Rows before cleaning: {rows_before}")
print(f"Rows after cleaning:  {rows_after}")
print(f"Rows dropped:         {rows_before - rows_after}")
```

```python
# Null-audit after cleaning (should be all zero on required fields)
timetable_clean_df.select([
    F.sum(F.col(c).isNull().cast("int")).alias(f"{c}_nulls")
    for c in ["Operator", "ServiceCode", "LineName", "Trips", "Stops"]
]).show()
```

## 7. Parquet Persistence

We persist the cleaned timetable to **Parquet**, partitioned by `Operator`. Parquet's columnar,
compressed format plus partition pruning means Notebook 2 can filter to a single operator's data
without scanning the whole dataset.

```python
timetable_parquet_path = os.path.join(PROCESSED_DIR, "timetable_parquet")

(
    timetable_clean_df
    .write
    .mode("overwrite")
    .partitionBy("Operator")
    .parquet(timetable_parquet_path)
)

# Verify round-trip
reloaded = spark.read.parquet(timetable_parquet_path)
print("Rows written & reloaded:", reloaded.count())
print("Partition folders on disk (sample):", os.listdir(timetable_parquet_path)[:5])
```

## 8. GTFS-Realtime Vehicle Locations

BODS vehicle-location feeds are published as **GTFS-Realtime protobuf** `FeedMessage` binaries
(the same `google.transit.realtime` schema used by transit agencies worldwide). We generate
snapshot `.bin` files using the official `gtfs-realtime-bindings` protobuf schema, so the parser
below reads exactly the wire format BODS serves.

```python
gtfs_pb2 = gtfs_realtime_pb2

def build_gtfs_rt_snapshot(snapshot_time, service_registry_slice):
    """Builds one real GTFS-RT FeedMessage (VehiclePosition entities) and returns serialized bytes."""
    feed = gtfs_pb2.FeedMessage()
    feed.header.gtfs_realtime_version = "2.0"
    feed.header.timestamp = int(snapshot_time.timestamp())

    entity_id = 0
    for svc in service_registry_slice:
        n_vehicles = max(1, svc["NumTrips"] // 10)  # busier services run more vehicles at once
        for v in range(n_vehicles):
            entity_id += 1
            entity = feed.entity.add()
            entity.id = f"{svc['OperatorID']}-VEH-{entity_id:05d}"
            vp = entity.vehicle
            vp.vehicle.id = entity.id
            vp.trip.trip_id = f"{svc['ServiceCode']}-T{v:03d}"
            vp.trip.route_id = svc["LineName"]
            # Great Britain bounding box for plausible coordinates
            vp.position.latitude = float(np.clip(np.random.normal(53.5, 2.0), 50.0, 58.5))
            vp.position.longitude = float(np.clip(np.random.normal(-2.0, 2.5), -6.0, 1.5))
            vp.position.bearing = float(random.randint(0, 359))
            vp.position.speed = float(np.clip(np.random.normal(22, 10), 0, 60))  # mph-ish
            vp.timestamp = int(snapshot_time.timestamp())

    return feed.SerializeToString()
```

```python
# Generate a rolling window of snapshots (simulates a live feed polled every 5 minutes)
if os.path.exists(VEHICLE_DIR):
    shutil.rmtree(VEHICLE_DIR)
os.makedirs(VEHICLE_DIR, exist_ok=True)

base_time = dt.datetime(2026, 7, 27, 6, 0, 0)
NUM_SNAPSHOTS = 6

for snap in range(NUM_SNAPSHOTS):
    snapshot_time = base_time + dt.timedelta(minutes=5 * snap)
    payload = build_gtfs_rt_snapshot(snapshot_time, service_registry)
    out_path = os.path.join(VEHICLE_DIR, f"vehicle_positions_{snap:02d}.bin")
    with open(out_path, "wb") as f:
        f.write(payload)

bin_files = glob.glob(os.path.join(VEHICLE_DIR, "*.bin"))
print(f"Generated {len(bin_files)} GTFS-Realtime snapshot files "
      f"({sum(os.path.getsize(p) for p in bin_files)/1024:.1f} KB total).")
```

### Parsing GTFS-Realtime with PySpark

Same distributed pattern as before: distribute the *file paths*, deserialize each protobuf
`FeedMessage` on the executor, and flatten every `VehiclePosition` entity into a row.

```python
def parse_gtfs_rt_file(file_path):
    feed = gtfs_pb2.FeedMessage()
    with open(file_path, "rb") as f:
        feed.ParseFromString(f.read())

    rows = []
    for entity in feed.entity:
        if entity.HasField("vehicle"):
            vp = entity.vehicle
            rows.append({
                "VehicleID": vp.vehicle.id,
                "TripID": vp.trip.trip_id,
                "RouteID": vp.trip.route_id,
                "Latitude": vp.position.latitude,
                "Longitude": vp.position.longitude,
                "Bearing": vp.position.bearing,
                "Speed": vp.position.speed,
                "Timestamp": vp.timestamp,
            })
    return rows

bin_paths_rdd = spark.sparkContext.parallelize(bin_files, numSlices=min(6, len(bin_files)))
vehicle_rows_rdd = bin_paths_rdd.flatMap(parse_gtfs_rt_file)

vehicle_schema = T.StructType([
    T.StructField("VehicleID", T.StringType(), True),
    T.StructField("TripID", T.StringType(), True),
    T.StructField("RouteID", T.StringType(), True),
    T.StructField("Latitude", T.DoubleType(), True),
    T.StructField("Longitude", T.DoubleType(), True),
    T.StructField("Bearing", T.DoubleType(), True),
    T.StructField("Speed", T.DoubleType(), True),
    T.StructField("Timestamp", T.LongType(), True),
])

vehicle_df = spark.createDataFrame(vehicle_rows_rdd, schema=vehicle_schema)
vehicle_df = vehicle_df.withColumn("EventTime", F.to_timestamp(F.from_unixtime("Timestamp")))
print("Vehicle-position pings parsed:", vehicle_df.count())
vehicle_df.show(5)
```

### Broadcast Join: attaching Operator identity to each vehicle ping

`TripID` encodes the `ServiceCode`, but the mapping from `ServiceCode → Operator` lives in the
(small) timetable summary — a perfect case for a **broadcast join**: Spark ships the small lookup
table to every executor instead of shuffling the much larger vehicle-ping table across the
network.

```python
# Build the small lookup table (one row per ServiceCode) — small enough to broadcast
service_lookup_df = (
    timetable_clean_df
    .select("ServiceCode", "Operator", "LineName")
    .withColumnRenamed("LineName", "TT_LineName")
    .dropDuplicates(["ServiceCode"])
)
print("Lookup table rows (broadcast candidate):", service_lookup_df.count())

vehicle_df = vehicle_df.withColumn("ServiceCode", F.regexp_extract("TripID", r"^(OP\d{3}-\d{2})", 1))

vehicle_enriched_df = vehicle_df.join(
    F.broadcast(service_lookup_df),
    on="ServiceCode",
    how="left",
)

print("\nPhysical plan (look for BroadcastHashJoin / BroadcastExchange below):\n")
vehicle_enriched_df.explain(mode="simple")
```

```python
vehicle_enriched_df = vehicle_enriched_df.persist(StorageLevel.MEMORY_AND_DISK)
print("Enriched vehicle pings:", vehicle_enriched_df.count())
vehicle_enriched_df.select("VehicleID", "Operator", "RouteID", "Latitude", "Longitude", "Speed").show(5)

# Null-audit: any pings that failed to match an Operator (data-quality signal worth tracking)
unmatched = vehicle_enriched_df.filter(F.col("Operator").isNull()).count()
print(f"Vehicle pings with no matching Operator: {unmatched} "
      f"({unmatched / vehicle_enriched_df.count():.1%})")
```

```python
vehicle_parquet_path = os.path.join(PROCESSED_DIR, "vehicle_locations_parquet")

(
    vehicle_enriched_df
    .filter(F.col("Operator").isNotNull())
    .write.mode("overwrite")
    .partitionBy("Operator")
    .parquet(vehicle_parquet_path)
)
print("Vehicle location parquet written to:", os.path.relpath(vehicle_parquet_path, PROJECT_ROOT))
```

## 9. SIRI-SX Disruptions

SIRI-SX (Service Interchange for Real-time Information — Situation eXchange) is the European
standard XML format BODS uses to publish disruption/incident data. We generate disruption
records weighted so that busier/larger operators (more services, more vehicles on the road)
have proportionally more opportunities for disruption — a realistic operational signal that
later becomes the `DisruptionRate` feature in Notebook 2.

```python
SEVERITIES = ["slight", "normal", "severe", "verySevere"]
DISRUPTION_TYPES = ["roadworks", "roadClosure", "vehicleBreakdown", "staffShortage",
                     "weather", "specialEvent", "publicDisorder"]

def build_siri_sx_xml(operator, operator_id, line_name, severity, disruption_type, start, end, description):
    ns = "http://www.siri.org.uk/siri"
    return f"""<?xml version="1.0" encoding="UTF-8"?>
<PtSituationElement xmlns="{ns}">
  <ParticipantRef>{operator_id}</ParticipantRef>
  <SituationNumber>{operator_id}-DIS-{random.randint(1000,9999)}</SituationNumber>
  <Severity>{severity}</Severity>
  <ReasonType>{disruption_type}</ReasonType>
  <Description>{description}</Description>
  <Affects><Operators><AffectedOperator><OperatorRef>{operator_id}</OperatorRef>
  <OperatorName>{operator}</OperatorName></AffectedOperator></Operators>
  <Lines><AffectedLine><LineRef>{line_name}</LineRef></AffectedLine></Lines></Affects>
  <ValidityPeriod><StartTime>{start.isoformat()}</StartTime><EndTime>{end.isoformat()}</EndTime></ValidityPeriod>
</PtSituationElement>"""
```

```python
if os.path.exists(DISRUPTION_DIR):
    shutil.rmtree(DISRUPTION_DIR)
os.makedirs(DISRUPTION_DIR, exist_ok=True)

disruption_file_count = 0
for svc in service_registry:
    # More trips -> more exposure -> higher chance of a logged disruption
    n_disruptions = np.random.poisson(lam=svc["NumTrips"] / 25)
    for d in range(n_disruptions):
        severity = random.choices(SEVERITIES, weights=[0.4, 0.35, 0.18, 0.07])[0]
        disruption_type = random.choice(DISRUPTION_TYPES)
        start = base_time + dt.timedelta(hours=random.randint(0, 240))
        end = start + dt.timedelta(minutes=random.randint(15, 240))
        description = f"{disruption_type} affecting {svc['LineName']} ({severity})"

        xml_str = build_siri_sx_xml(
            svc["Operator"], svc["OperatorID"], svc["LineName"],
            severity, disruption_type, start, end, description
        )
        fname = f"{svc['OperatorID']}_{svc['ServiceCode']}_{d}.xml"
        with open(os.path.join(DISRUPTION_DIR, fname), "w", encoding="utf-8") as f:
            f.write(xml_str)
        disruption_file_count += 1

print(f"Generated {disruption_file_count} SIRI-SX disruption XML files.")
```

```python
def parse_siri_sx_file(file_path):
    ns = {"s": "http://www.siri.org.uk/siri"}
    try:
        root = ET.parse(file_path).getroot()
        return [{
            "Operator": root.find(".//s:AffectedOperator/s:OperatorName", ns).text,
            "Line": root.find(".//s:AffectedLine/s:LineRef", ns).text,
            "Severity": root.find("s:Severity", ns).text,
            "DisruptionType": root.find("s:ReasonType", ns).text,
            "Description": root.find("s:Description", ns).text,
            "StartTime": root.find(".//s:ValidityPeriod/s:StartTime", ns).text,
            "EndTime": root.find(".//s:ValidityPeriod/s:EndTime", ns).text,
        }]
    except Exception as e:
        return [{"Operator": None, "Line": None, "Severity": None, "DisruptionType": None,
                  "Description": f"PARSE_ERROR: {e}", "StartTime": None, "EndTime": None}]

disruption_files = glob.glob(os.path.join(DISRUPTION_DIR, "*.xml"))
disruption_paths_rdd = spark.sparkContext.parallelize(disruption_files, numSlices=min(6, len(disruption_files)))
disruption_rows_rdd = disruption_paths_rdd.flatMap(parse_siri_sx_file)

disruption_schema = T.StructType([
    T.StructField("Operator", T.StringType(), True),
    T.StructField("Line", T.StringType(), True),
    T.StructField("Severity", T.StringType(), True),
    T.StructField("DisruptionType", T.StringType(), True),
    T.StructField("Description", T.StringType(), True),
    T.StructField("StartTime", T.StringType(), True),
    T.StructField("EndTime", T.StringType(), True),
])

disruption_df = spark.createDataFrame(disruption_rows_rdd, schema=disruption_schema)
disruption_df = (
    disruption_df
    .filter(F.col("Operator").isNotNull())
    .withColumn("StartTime", F.to_timestamp("StartTime"))
    .withColumn("EndTime", F.to_timestamp("EndTime"))
    .withColumn("DurationMinutes", (F.col("EndTime").cast("long") - F.col("StartTime").cast("long")) / 60)
    .persist(StorageLevel.MEMORY_AND_DISK)
)
print("Disruption records parsed & cleaned:", disruption_df.count())
disruption_df.show(5, truncate=40)
```

```python
disruption_parquet_path = os.path.join(PROCESSED_DIR, "disruptions_parquet")
(
    disruption_df.write.mode("overwrite")
    .partitionBy("Operator")
    .parquet(disruption_parquet_path)
)
print("Disruption parquet written to:", os.path.relpath(disruption_parquet_path, PROJECT_ROOT))
```

## 10. Spark-SQL Summary Tables

We now register temp views and build the three summary tables required, purely in **Spark SQL**,
then join them into a single operator-level master table.

```python
timetable_clean_df.createOrReplaceTempView("timetable")
vehicle_enriched_df.createOrReplaceTempView("vehicles")
disruption_df.createOrReplaceTempView("disruptions")

operator_summary_df = spark.sql("""
    SELECT
        Operator,
        COUNT(*)              AS num_services,
        SUM(Trips)             AS total_trips,
        AVG(Trips)              AS avg_trips_per_service,
        SUM(Stops)             AS total_stop_visits,
        AVG(Stops)               AS avg_stops_per_service
    FROM timetable
    GROUP BY Operator
""")
operator_summary_df.show(10, truncate=False)
```

```python
vehicle_summary_df = spark.sql("""
    SELECT
        Operator,
        COUNT(*)                AS num_pings,
        COUNT(DISTINCT VehicleID) AS distinct_vehicles,
        AVG(Speed)                 AS avg_speed,
        STDDEV(Speed)                AS speed_stddev
    FROM vehicles
    WHERE Operator IS NOT NULL
    GROUP BY Operator
""")
vehicle_summary_df.show(10, truncate=False)
```

```python
disruption_summary_df = spark.sql("""
    SELECT
        Operator,
        COUNT(*)                                        AS num_disruptions,
        AVG(DurationMinutes)                               AS avg_disruption_minutes,
        SUM(CASE WHEN Severity IN ('severe','verySevere') THEN 1 ELSE 0 END) AS num_severe_disruptions
    FROM disruptions
    GROUP BY Operator
""")
disruption_summary_df.show(10, truncate=False)
```

```python
# Join all three summaries into a single operator master table.
# LEFT JOIN from operator_summary (every registered operator) so operators with
# zero disruptions or no vehicle pings still appear, with COALESCE to 0.
operator_summary_df.createOrReplaceTempView("op_summary")
vehicle_summary_df.createOrReplaceTempView("veh_summary")
disruption_summary_df.createOrReplaceTempView("dis_summary")

operator_master_df = spark.sql("""
    SELECT
        o.Operator,
        o.num_services, o.total_trips, o.avg_trips_per_service,
        o.total_stop_visits, o.avg_stops_per_service,
        COALESCE(v.num_pings, 0)          AS num_pings,
        COALESCE(v.distinct_vehicles, 0)  AS distinct_vehicles,
        COALESCE(v.avg_speed, 0)          AS avg_speed,
        COALESCE(v.speed_stddev, 0)       AS speed_stddev,
        COALESCE(d.num_disruptions, 0)          AS num_disruptions,
        COALESCE(d.avg_disruption_minutes, 0)   AS avg_disruption_minutes,
        COALESCE(d.num_severe_disruptions, 0)   AS num_severe_disruptions
    FROM op_summary o
    LEFT JOIN veh_summary v ON o.Operator = v.Operator
    LEFT JOIN dis_summary d ON o.Operator = d.Operator
""")

print("Physical plan for the three-way operator join (shuffle/broadcast strategy chosen by AQE):\n")
operator_master_df.explain(mode="simple")
```

```python
operator_master_df = operator_master_df.persist(StorageLevel.MEMORY_AND_DISK)
print("Operator master rows:", operator_master_df.count())
operator_master_df.orderBy(F.desc("total_trips")).show(10, truncate=False)
```

## 11. SQLite Relational Storage

The regulator needs a lightweight relational store for dashboarding/ad-hoc SQL outside Spark.
We define an explicit schema with primary/foreign keys, and **every** query below uses
**parameterized placeholders (`?`)** — never raw string interpolation — to eliminate SQL-injection
risk, per the assignment's security requirement.

```python
DB_PATH = os.path.join(SQLITE_DIR, "bods_operator_benchmark.db")
if os.path.exists(DB_PATH):
    os.remove(DB_PATH)

conn = sqlite3.connect(DB_PATH)
cur = conn.cursor()

cur.executescript("""
    PRAGMA foreign_keys = ON;

    CREATE TABLE operators (
        operator_id     TEXT PRIMARY KEY,
        operator_name   TEXT NOT NULL UNIQUE
    );

    CREATE TABLE operator_summary (
        operator_name           TEXT PRIMARY KEY,
        num_services             INTEGER,
        total_trips                INTEGER,
        avg_trips_per_service        REAL,
        total_stop_visits          INTEGER,
        avg_stops_per_service        REAL,
        num_pings                   INTEGER,
        distinct_vehicles             INTEGER,
        avg_speed                       REAL,
        speed_stddev                      REAL,
        num_disruptions                     INTEGER,
        avg_disruption_minutes                REAL,
        num_severe_disruptions                  INTEGER,
        FOREIGN KEY (operator_name) REFERENCES operators(operator_name)
    );

    CREATE TABLE disruption_events (
        disruption_id   INTEGER PRIMARY KEY AUTOINCREMENT,
        operator_name   TEXT NOT NULL,
        line            TEXT,
        severity        TEXT,
        disruption_type TEXT,
        start_time      TEXT,
        end_time        TEXT,
        duration_minutes REAL,
        FOREIGN KEY (operator_name) REFERENCES operators(operator_name)
    );
""")
conn.commit()
print("SQLite schema created at:", os.path.relpath(DB_PATH, PROJECT_ROOT))
```

```python
# ----------------------------------------------------------------------
# Insert data using PARAMETERIZED queries (executemany with "?" placeholders).
# We convert the small, already-aggregated Spark DataFrames to pandas for the
# insert step only — the heavy lifting (parsing, joining, aggregating) all
# happened in Spark above.
# ----------------------------------------------------------------------
operator_ids_df = pd.DataFrame(
    [(OPERATOR_IDS[name], name) for name in OPERATOR_IDS], columns=["operator_id", "operator_name"]
)
cur.executemany(
    "INSERT INTO operators (operator_id, operator_name) VALUES (?, ?)",
    list(operator_ids_df.itertuples(index=False, name=None)),
)

master_pd = operator_master_df.toPandas()
summary_cols = [
    "Operator", "num_services", "total_trips", "avg_trips_per_service",
    "total_stop_visits", "avg_stops_per_service", "num_pings", "distinct_vehicles",
    "avg_speed", "speed_stddev", "num_disruptions", "avg_disruption_minutes",
    "num_severe_disruptions",
]
cur.executemany(
    f"INSERT INTO operator_summary ({', '.join(['operator_name']+summary_cols[1:])}) "
    f"VALUES ({', '.join(['?'] * len(summary_cols))})",
    list(master_pd[summary_cols].itertuples(index=False, name=None)),
)

disruption_events_pd = disruption_df.select(
    "Operator", "Line", "Severity", "DisruptionType", "StartTime", "EndTime", "DurationMinutes"
).toPandas()
disruption_events_pd["StartTime"] = disruption_events_pd["StartTime"].astype(str)
disruption_events_pd["EndTime"] = disruption_events_pd["EndTime"].astype(str)
cur.executemany(
    """INSERT INTO disruption_events
       (operator_name, line, severity, disruption_type, start_time, end_time, duration_minutes)
       VALUES (?, ?, ?, ?, ?, ?, ?)""",
    list(disruption_events_pd.itertuples(index=False, name=None)),
)

conn.commit()
print("Rows inserted -> operators:", len(operator_ids_df),
      "| operator_summary:", len(master_pd),
      "| disruption_events:", len(disruption_events_pd))
```

```python
# Example parameterized SELECT (safe against SQL injection — never f-string a WHERE clause)
def get_operator_summary(operator_name: str) -> pd.DataFrame:
    query = "SELECT * FROM operator_summary WHERE operator_name = ?"
    return pd.read_sql_query(query, conn, params=(operator_name,))

def get_top_n_by_disruptions(n: int) -> pd.DataFrame:
    query = """
        SELECT operator_name, num_disruptions, num_severe_disruptions
        FROM operator_summary
        ORDER BY num_disruptions DESC
        LIMIT ?
    """
    return pd.read_sql_query(query, conn, params=(n,))

print(get_operator_summary("Stagecoach North East"))
print()
print(get_top_n_by_disruptions(5))
```

## 12. Data Validation & Schema Checking

```python
# ----------------------------------------------------------------------
# Schema validation: assert the persisted Parquet matches the expected contract
# ----------------------------------------------------------------------
expected_timetable_fields = {"Operator", "ServiceCode", "LineName", "Description", "Trips", "Stops", "XML_File"}
reloaded_timetable = spark.read.parquet(timetable_parquet_path)
actual_fields = set(reloaded_timetable.columns)

assert expected_timetable_fields.issubset(actual_fields), \
    f"Schema drift detected! Missing: {expected_timetable_fields - actual_fields}"
print("✅ Timetable schema contract satisfied.")
print("Schema:")
reloaded_timetable.printSchema()
```

```python
# ----------------------------------------------------------------------
# Null-rate / cardinality / duplicate audit across all three processed tables
# ----------------------------------------------------------------------
def validation_report(df, name, key_col):
    total = df.count()
    nulls = {c: df.filter(F.col(c).isNull()).count() for c in df.columns}
    dup_keys = total - df.select(key_col).distinct().count()
    cardinality = {c: df.select(c).distinct().count() for c in df.columns if df.schema[c].dataType.simpleString() == "string"}
    print(f"--- {name} ---")
    print(f"Row count: {total}")
    print(f"Duplicate {key_col} values: {dup_keys}")
    print("Null counts:", {k: v for k, v in nulls.items() if v > 0} or "none")
    print("Distinct-value cardinality (string cols):", cardinality)
    print()

validation_report(timetable_clean_df, "Timetable", "ServiceCode")
validation_report(vehicle_enriched_df, "Vehicle Locations", "VehicleID")
validation_report(disruption_df, "Disruptions", "Description")
```

```python
# ----------------------------------------------------------------------
# Outlier sanity checks on numeric ranges (defensive checks before ML in Notebook 3)
# ----------------------------------------------------------------------
bad_speed = vehicle_enriched_df.filter((F.col("Speed") < 0) | (F.col("Speed") > 80)).count()
bad_coords = vehicle_enriched_df.filter(
    (F.col("Latitude") < 49) | (F.col("Latitude") > 61) |
    (F.col("Longitude") < -8) | (F.col("Longitude") > 2)
).count()

print(f"Vehicle pings with implausible speed (<0 or >80 mph): {bad_speed}")
print(f"Vehicle pings outside GB bounding box: {bad_coords}")
assert bad_speed == 0 and bad_coords == 0, "Range validation failed — investigate before proceeding."
print("✅ Range validation passed.")
```

## 13. Checkpointing vs. Caching — Lineage Discussion

`persist()`/`cache()` stores computed partitions in memory/disk **but still keeps the full
lineage graph** (Spark can recompute from source if a cached partition is evicted). For a long
transformation chain like `operator_master_df` (three joins deep), that lineage graph grows large.
**`checkpoint()`** truncates the lineage entirely by writing the DataFrame to reliable storage
(`CHECKPOINT_DIR`, set on the SparkContext earlier) and starting a fresh lineage from that point —
useful in production pipelines with many iterative joins/loops, to bound recovery cost.

```python
print("RDD lineage depth BEFORE checkpoint (debug string length as a proxy):",
      len(operator_master_df.rdd.toDebugString().decode().splitlines()))

operator_master_checkpointed = operator_master_df.checkpoint(eager=True)

print("RDD lineage depth AFTER checkpoint:",
      len(operator_master_checkpointed.rdd.toDebugString().decode().splitlines()))
print("\n(The checkpointed DataFrame's lineage now starts from the checkpoint file on disk,")
print(" not from the three-way join + XML parse chain above.)")
```
