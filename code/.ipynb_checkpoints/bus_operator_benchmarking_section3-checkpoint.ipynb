# SECTION 3 – Feature Engineering Using PySpark

This section focuses on data cleaning, type casting, aggregating operator-level features (total trips, average stops, unique routes, etc.), running PySpark SQL queries, and exporting the results to Parquet and CSV formats.

### Cell 1 – Check Dataset

```python
services_df.printSchema()
services_df.show(5, truncate=False)
```

### Cell 2 – Count Total Records

```python
print("Total Services:", services_df.count())
```

### Cell 3 – Check Missing Values

```python
from pyspark.sql.functions import col, count, when

services_df.select([
    count(when(col(c).isNull(), c)).alias(c)
    for c in services_df.columns
]).show()
```

### Cell 4 – Remove Duplicate Records

```python
services_df = services_df.dropDuplicates()
print("Rows After Removing Duplicates:")
print(services_df.count())
```

### Cell 5 – Convert Trips and Stops to Integer

```python
from pyspark.sql.functions import col

services_df = services_df \
    .withColumn("Trips", col("Trips").cast("int")) \
    .withColumn("Stops", col("Stops").cast("int"))
```

### Cell 6 – Create Operator-Level Features

```python
from pyspark.sql.functions import *

operator_features = services_df.groupBy("Operator").agg(
    count("*").alias("TotalServices"),
    sum("Trips").alias("TotalTrips"),
    avg("Trips").alias("AverageTrips"),
    sum("Stops").alias("TotalStops"),
    avg("Stops").alias("AverageStops"),
    countDistinct("LineName").alias("UniqueRoutes"),
    countDistinct("ServiceCode").alias("UniqueServiceCodes")
)
```

### Cell 7 – View Features

```python
operator_features.show(20, truncate=False)
```

### Cell 8 – Sort by Largest Operators

```python
operator_features.orderBy(
    col("TotalTrips").desc()
).show(20, truncate=False)
```

### Cell 9 – Register Spark SQL Table

```python
operator_features.createOrReplaceTempView("operator_features")
```

### Cell 10 – Spark SQL Query

```python
spark.sql("""
SELECT
    Operator,
    TotalServices,
    TotalTrips,
    AverageTrips,
    TotalStops,
    AverageStops,
    UniqueRoutes
FROM operator_features
ORDER BY TotalTrips DESC
LIMIT 20
""").show(truncate=False)
```

### Cell 11 – Cache the Data

```python
operator_features.cache()
operator_features.count()
```

### Cell 12 – Check Number of Partitions

```python
print(operator_features.rdd.getNumPartitions())
```

### Cell 13 – Repartition

```python
operator_features = operator_features.repartition(8)
print(operator_features.rdd.getNumPartitions())
```

### Cell 14 – Save as Parquet

```python
operator_features.write.mode("overwrite").parquet("output/operator_features")
```

### Cell 15 – Save to CSV

```python
operator_features.toPandas().to_csv(
    "output/operator_features.csv",
    index=False
)
```
