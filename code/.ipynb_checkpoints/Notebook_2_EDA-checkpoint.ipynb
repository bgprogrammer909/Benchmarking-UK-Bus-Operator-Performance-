# Notebook 2 — Exploratory Data Analysis & Distributed Profiling
This notebook demonstrates PySpark SQL queries, statistical profiling (mean, median, std, skewness, kurtosis), quality assessment, and visualization.

```python
# SECTION 1 — Load & Verify Distributed Parquet Data
import os
import sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pyspark.sql import SparkSession
from pyspark.sql.functions import *

os.environ["PYSPARK_PYTHON"] = sys.executable
os.environ["PYSPARK_DRIVER_PYTHON"] = sys.executable

spark = (
    SparkSession.builder
    .appName("Bus Service Benchmarking - EDA")
    .config("spark.sql.shuffle.partitions", "16")
    .config("spark.driver.memory", "6g")
    .getOrCreate()
)
spark.sparkContext.setLogLevel("WARN")

df = spark.read.parquet("output/trips_master_parquet").cache()
num_records = df.count()
print(f"Successfully loaded {num_records} Trip Records for EDA.")
print(f"Partition Count: {df.rdd.getNumPartitions()}")
```

```python
# SECTION 2 — Data Profiling & Statistical Measures
print("\n--- Null Count Assessment ---")
df.select([count(when(col(c).isNull(), c)).alias(c) for c in df.columns]).show()

print("--- Statistical Metrics (Mean, Median, Std, Skewness, Kurtosis) ---")
df.select(
    mean("ActualDelayMin").alias("Mean_Delay"),
    expr("percentile_approx(ActualDelayMin, 0.5)").alias("Median_Delay"),
    stddev("ActualDelayMin").alias("Std_Delay"),
    skewness("ActualDelayMin").alias("Skew_Delay"),
    kurtosis("ActualDelayMin").alias("Kurtosis_Delay")
).show()

df.select(
    mean("ServicePerformanceIndex").alias("Mean_SPI"),
    expr("percentile_approx(ServicePerformanceIndex, 0.5)").alias("Median_SPI"),
    stddev("ServicePerformanceIndex").alias("Std_SPI"),
    skewness("ServicePerformanceIndex").alias("Skew_SPI"),
    kurtosis("ServicePerformanceIndex").alias("Kurtosis_SPI")
).show()
```

```python
# SECTION 3 — PySpark SQL Aggregations & Visualisations
df.createOrReplaceTempView("trips")

print("--- Delay by Hour of Day ---")
hourly_df = spark.sql("""
    SELECT DepartureHour, COUNT(*) as TripCount, AVG(ActualDelayMin) as AvgDelay
    FROM trips
    GROUP BY DepartureHour
    ORDER BY DepartureHour
""").toPandas()

plt.figure(figsize=(10, 4))
sns.barplot(data=hourly_df, x="DepartureHour", y="AvgDelay", palette="magma")
plt.title("Average Bus Delay by Hour of Departure")
plt.xlabel("Hour of Day")
plt.ylabel("Average Delay (Minutes)")
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()

# Correlation Matrix (Converted to Pandas only for plotting)
sample_pdf = df.select("StopCount", "AvgSpeed", "DisruptionPenalty", "ActualDelayMin", "ServicePerformanceIndex") \
    .sample(fraction=0.1, seed=42).toPandas()

plt.figure(figsize=(8, 5))
sns.heatmap(sample_pdf.corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Feature Correlation Matrix")
plt.tight_layout()
plt.show()
```
