This section sets up the required environment, initializes a Apache Spark session, extracts compressed timetable files, and summarizes the dataset across operators.

### Cell 2 — Import Libraries

```python
import os
import glob
import zipfile
import xml.etree.ElementTree as ET
from pathlib import Path
from tqdm import tqdm

import pandas as pd

from pyspark.sql import SparkSession
from pyspark.sql.types import *
from pyspark.sql.functions import *
```

### Cell 3 — Create Spark Session

```python
spark = (
    SparkSession.builder
    .appName("Bus Operator Benchmarking")
    .config("spark.sql.shuffle.partitions", "8")
    .config("spark.driver.memory", "4g")
    .config("spark.executor.memory", "4g")
    .getOrCreate()
)

spark.sparkContext.setLogLevel("WARN")

print("Spark Version :", spark.version)
```

### Cell 4 — Dataset Location & Automatic ZIP Extraction
Update `BASE_PATH` to point to your dataset directory. This block automatically extracts any nested `.zip` files.

```python
BASE_PATH = r"D:\big data assgn\bodds_archive_20260727"

# Automatically Extract Every ZIP
zip_files = list(Path(BASE_PATH).rglob("*.zip"))
print(f"ZIP files found : {len(zip_files)}")

for file in tqdm(zip_files):
    try:
        with zipfile.ZipFile(file, 'r') as zip_ref:
            zip_ref.extractall(file.parent)
    except Exception as e:
        print(file)
```

### Cell 5 — Find Every XML File

```python
xml_files = list(Path(BASE_PATH).rglob("*.xml"))
print("Total XML Files :", len(xml_files))
```

### Cell 6 — Show Sample Files

```python
for file in xml_files[:10]:
    print(file)
```

### Cell 7 — Count Operators & Convert to Pandas DataFrame

```python
operators = {}
for file in xml_files:
    operator = file.parent.name
    operators[operator] = operators.get(operator, 0) + 1

# Convert to Pandas
operator_df = pd.DataFrame(
    operators.items(),
    columns=[
        "Operator",
        "Timetable Files"
    ]
)

operator_df.head()
```

### Cell 8 — Convert to PySpark DataFrame & Count

```python
operator_spark = spark.createDataFrame(operator_df)
operator_spark.show(10)

print("Operators :", operator_spark.count())
```

### Cell 9 — Spark SQL Query
Query top 20 operators ordered by timetable file count.

```python
operator_spark.createOrReplaceTempView("operators")
spark.sql("""
SELECT *
FROM operators
ORDER BY `Timetable Files` DESC
LIMIT 20
""").show()
```

### Cell 10 — Partition Management & Caching
Check initial partitions, repartition to 8 partitions, and cache the dataframe.

```python
print("Initial partitions:", operator_spark.rdd.getNumPartitions())

# Repartition
operator_spark = operator_spark.repartition(8)
print("Repartitioned count:", operator_spark.rdd.getNumPartitions())

# Cache
operator_spark.cache()
operator_spark.count()
```
